# Laboratorio SQL

Consultas exploratorias sobre el histórico de Supabase. **Es de solo lectura:** cada consulta corre dentro de una transacción `READ ONLY`, así que un `UPDATE` o un `DELETE` escrito acá falla en vez de tocar la base.

No es parte de la corrida diaria. Para el reporte está `pipeline.py`; este notebook solo lee lo que la corrida ya guardó.

```python
# Como en el notebook principal: subir a la raíz del proyecto (donde está pipeline.py)
import os
import sys
from pathlib import Path

RAIZ = Path.cwd()
if not (RAIZ / "pipeline.py").exists():
    RAIZ = RAIZ.parent
if not (RAIZ / "pipeline.py").exists():
    raise RuntimeError(f"No encuentro pipeline.py desde {Path.cwd()}")
os.chdir(RAIZ)
if str(RAIZ) not in sys.path:
    sys.path.insert(0, str(RAIZ))

import pandas as pd
from sqlalchemy import text

from reporte import data_access

engine = data_access.crear_engine()


def consulta(sql: str, **params) -> pd.DataFrame:
    """Corre `sql` en una transacción de solo lectura y devuelve un DataFrame."""
    with engine.connect() as conn:
        conn.execute(text("SET TRANSACTION READ ONLY"))
        return pd.read_sql_query(text(sql), conn, params=params)


consulta('SELECT COUNT(*) AS filas, MIN("Fecha") AS desde, MAX("Fecha") AS hasta FROM "Fact_Mercado_Macro"')
```

## Días hábiles sin fila

Lunes a viernes de los últimos 90 días que no tienen fila. Incluye los feriados (la tabla no sabe cuáles son), así que hay que mirar la lista con el calendario al lado: lo que no sea feriado es un día en que la corrida no se hizo.

```python
DIAS = ["lunes", "martes", "miércoles", "jueves", "viernes"]
faltantes = consulta('''
    SELECT d::date AS fecha, extract(isodow FROM d)::int AS dia_semana
    FROM generate_series(CURRENT_DATE - 90, CURRENT_DATE - 1, interval '1 day') AS d
    WHERE extract(isodow FROM d) < 6
      AND NOT EXISTS (SELECT 1 FROM "Fact_Mercado_Macro" f WHERE f."Fecha" = d::date)
    ORDER BY d
''')
faltantes.assign(dia=faltantes["dia_semana"].map(lambda n: DIAS[n - 1])).drop(columns="dia_semana")
```

## Brecha entre el blue y el MEP, por mes

Promedio, mínimo y máximo de la brecha diaria. Positiva: el blue está más caro que el MEP.

```python
brecha = consulta('''
    SELECT date_trunc('month', "Fecha")::date AS mes,
           ROUND(AVG(("TCV_Blue" / "TCV_MEP" - 1) * 100)::numeric, 2) AS brecha_promedio,
           ROUND(MIN(("TCV_Blue" / "TCV_MEP" - 1) * 100)::numeric, 2) AS brecha_minima,
           ROUND(MAX(("TCV_Blue" / "TCV_MEP" - 1) * 100)::numeric, 2) AS brecha_maxima,
           COUNT(*) AS ruedas
    FROM "Fact_Mercado_Macro"
    WHERE "Fecha" >= :desde AND "TCV_MEP" > 0
    GROUP BY 1
    ORDER BY 1
''', desde="2025-01-01")
brecha
```

```python
brecha.plot(x="mes", y=["brecha_promedio", "brecha_minima", "brecha_maxima"], figsize=(10, 4),
            title="Brecha blue contra MEP (%)", grid=True);
```

## Riesgo país por mes

```python
consulta('''
    SELECT date_trunc('month', "Fecha")::date AS mes,
           MIN(riesgo_pais) AS minimo, MAX(riesgo_pais) AS maximo,
           ROUND(AVG(riesgo_pais)::numeric, 0) AS promedio,
           (array_agg(riesgo_pais ORDER BY "Fecha" DESC))[1] AS cierre
    FROM "Fact_Mercado_Macro"
    WHERE "Fecha" >= :desde
    GROUP BY 1
    ORDER BY 1
''', desde="2025-01-01")
```

## Variación mensual de blue, MEP y billete

Último dato de cada mes contra el último del mes anterior.

```python
cierres = consulta('''
    SELECT DISTINCT ON (date_trunc('month', "Fecha"))
           date_trunc('month', "Fecha")::date AS mes, "TCV_Blue", "TCV_MEP", "TCV_Billete"
    FROM "Fact_Mercado_Macro"
    WHERE "Fecha" >= :desde
    ORDER BY date_trunc('month', "Fecha"), "Fecha" DESC
''', desde="2024-12-01").set_index("mes")
(cierres.pct_change() * 100).round(2).dropna()
```

## Párrafos de IA: qué modelo los escribió

Cuántos párrafos salieron de cada modelo y cuántos días quedaron sin párrafo (`ai_paragraph` vacío o nulo).

```python
consulta('''
    SELECT COALESCE(ai_model, '(sin modelo)') AS modelo,
           COUNT(*) AS dias,
           COUNT(*) FILTER (WHERE COALESCE(ai_paragraph, '') = '') AS sin_parrafo,
           ROUND(AVG(length(ai_paragraph))) AS largo_promedio,
           MIN("Fecha") AS primero, MAX("Fecha") AS ultimo
    FROM "Fact_Mercado_Macro"
    WHERE "Fecha" >= :desde
    GROUP BY 1
    ORDER BY dias DESC
''', desde="2026-01-01")
```

## Auditoría del histórico: errores de escala y carga

**SQL + pandas, sin nuevas dependencias.** SQL lee la tabla en una transacción `READ ONLY`; pandas cruza fechas y fuentes y reutiliza las reglas de `scripts/depurar_historico.py`, las mismas de la limpieza anterior. Para este volumen no hace falta Polars ni DuckDB: sumarían otro motor sin mejorar la auditoría. Polars serviría para transformaciones sobre millones de filas; DuckDB para consultar archivos CSV/Parquet grandes con SQL sin cargarlos completos en pandas.

Busca comas corridas (por 10, 100, 1000 y otras potencias, incluso valores divididos), ceros perdidos, fechas con día y mes invertidos, rellenos repetidos, cotizaciones estancadas, compras mayores que ventas, compras faltantes y tasas inconsistentes. Cruza BNA, DolarHoy, Ámbito, ArgentinaDatos, BCRA y la FED con las tolerancias del script. Una diferencia de fuente por sí sola no demuestra un error.

**No aplica cambios.** `despues` es una propuesta calculada en memoria; un vacío propuesto significa que no hay reemplazo confiable. Puede haber varias reglas sobre una celda: se muestran todas, en su orden. Un informe vacío no garantiza que todos los valores sean correctos: depende de las reglas y la cobertura de las fuentes.

La descarga de referencias usa internet, puede tardar y falla si una fuente no responde. No modifica Supabase ni envía correos. Ejecutá estas celdas cuando quieras auditar, no como parte del reporte diario. No publiques resultados que no quieras compartir: quedan guardados en el notebook si lo guardás con salidas.


```python
from scripts import depurar_historico as depuracion

# consulta() ya impone READ ONLY; no se llama a aplicar() ni al CLI del script.
tabla_auditada = consulta('SELECT * FROM "Fact_Mercado_Macro" ORDER BY "Fecha"')
tabla_auditada["Fecha"] = pd.to_datetime(tabla_auditada["Fecha"], format="%Y-%m-%d")
tabla_auditada = tabla_auditada.set_index("Fecha")
if tabla_auditada.empty:
    raise ValueError("No hay filas para auditar.")
referencias_auditoria = depuracion.bajar_referencias(tabla_auditada.index.min().strftime("%Y-%m-%d"))
_, hallazgos = depuracion.correcciones(tabla_auditada, referencias_auditoria)
print(f"{len(tabla_auditada):,} filas revisadas; "
      f"{len(hallazgos.drop_duplicates(['Fecha', 'columna'])):,} celdas con propuestas.")
```

```python
if hallazgos.empty:
    print("Sin propuestas con estas reglas y las referencias disponibles.")
else:
    display(hallazgos.groupby(["columna", "motivo"]).agg(
        propuestas=("Fecha", "size"), desde=("Fecha", "min"), hasta=("Fecha", "max")
    ).reset_index())
    display(hallazgos.sort_values(["Fecha", "columna"], kind="stable"))
```

### Revisar solo errores de escala

El factor es `antes / despues`: 10, 100 o 1000 indica el exceso de escala; 0,1 indica un valor diez veces menor. Un mismo valor puede pasar por más de una regla. No confundas estos factores con variaciones de mercado.


```python
errores_escala = hallazgos.loc[
    hallazgos["motivo"].str.contains("coma decimal corrida", regex=False)
].copy()
errores_escala["factor_antes_sobre_despues"] = (
    errores_escala["antes"] / errores_escala["despues"].replace(0, float("nan"))
)
display(errores_escala)
```
