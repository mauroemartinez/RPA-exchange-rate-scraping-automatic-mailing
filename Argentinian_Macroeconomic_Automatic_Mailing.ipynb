## Consumo de APIs

```python
# Configuración general
def get_bcra_variable(id_variable: int) -> pd.DataFrame:
    url = f"{bcra_api_url}/{id_variable}"
    response = requests.get(
        url,
        timeout=30,
        verify=False
    )
    response.raise_for_status()

    return pd.DataFrame(response.json()["results"][0]["detalle"])

# BADLAR (140)

bcra_tea = get_bcra_variable(140)
bcra_tea = float(bcra_tea.iloc[-1]["valor"])

# Inflación (27)

inflacion_df = get_bcra_variable(27)
inflacion_df.columns = ["Fecha", "Inflación Mensual"]

# Fecha
inflacion_df["Fecha"] = pd.to_datetime(inflacion_df["Fecha"])

# Orden cronológico (para rolling)
inflacion_df = inflacion_df.iloc[::-1]

# Cálculos acumulados

factor = inflacion_df["Inflación Mensual"] / 100 + 1

inflacion_df["Inflación Bimestral"] = (
    factor.rolling(2).apply(np.prod, raw=True) - 1
) * 100

inflacion_df["Inflación Trimestral"] = (
    factor.rolling(3).apply(np.prod, raw=True) - 1
) * 100

inflacion_df["Inflación Anual"] = (
    factor.rolling(12, min_periods=12).apply(np.prod, raw=True) - 1
) * 100

inflacion_df["bcra_tea"] = bcra_tea

# Output

inflacion_df = (
    inflacion_df
    .iloc[::-1]
    .head(12)
)

inflacion_df["Fecha"] = inflacion_df["Fecha"].dt.strftime("%d-%m-%Y")

display(
    inflacion_df.style.format({
        "Inflación Mensual": "{:.2f}%",
        "Inflación Bimestral": "{:.2f}%",
        "Inflación Trimestral": "{:.2f}%",
        "Inflación Anual": "{:.2f}%",
        "bcra_tea": "{:.2f}%"
    })
)
```

## 💾 Exportar a CSV y Actualizar fila en SQL

```python
# Usamos SQLAlchemy para actualizar la neuva fila en SQL Server
# Configuración de conexión 
params = urllib.parse.quote_plus(
    r'DRIVER={ODBC Driver 18 for SQL Server};'
    r'SERVER=localhost\SQLEXPRESS;'
    r'DATABASE=MacroeconomicAnalytics;'
    r'Trusted_Connection=yes;'
)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# Definimos las columnas a exportar, que deben ser exactamente las mismas que las de la tabla en SQL Server (en orden y tipo de dato)
columnas_a_exportar = [
    'Fecha', 'TCC_Blue', 'TCV_Blue', 'TCC_Billete', 'TCV_Billete', 
    'TCC_Divisas', 'TCV_Divisas', 'Solidario', 'TCV_MEP', 'riesgo_pais', 
    'TCC_Euro', 'TCV_Euro', 'fed_tea', 'bcra_tea'
]
# Exportamos a csv como backup, y luego a SQL Server. De esta forma, si SQL falla, el CSV ya quedó actualizado.
df[columnas_a_exportar].to_csv(RUTA_BBDD, index=False, encoding="latin1")
print("✅ Backup CSV actualizado correctamente.")

# Carga a SQL Server con control de duplicados por fecha
try:
    # A. Traemos las fechas que ya están en SQL para comparar en formato fecha y como lista
    query_fechas = "SELECT Fecha FROM Fact_Mercado_Macro"
    fechas_db = pd.read_sql(query_fechas, con=engine)
    fechas_db_list = pd.to_datetime(fechas_db['Fecha']).dt.date.tolist()
    
    # B. Filtramos el DataFrame actual: solo filas cuya fecha NO esté en SQL
    df['Fecha'] = pd.to_datetime(df['Fecha']).dt.date
    df_nuevos_datos = df[~df['Fecha'].isin(fechas_db_list)]

    # C. Si hay datos nuevos, los "pusheamos"
    if not df_nuevos_datos.empty:
        df_nuevos_datos[columnas_a_exportar].to_sql(
            name='Fact_Mercado_Macro', 
            con=engine, 
            if_exists='append', 
            index=False
        )
        print(f"🚀 SQL Server: Se insertaron {len(df_nuevos_datos)} registros nuevos.")
    else:
        print("ℹ️ SQL Server: La base de datos ya está al día. No se insertó nada.")
except Exception as e:
    # Si SQL falla (por ejemplo, si no tenés WiFi o el servidor está caído), el script no se rompe y te avisa, pero el CSV ya quedó guardado arriba.
    print(f"⚠️ Error al conectar con SQL Server: {e}")
df
```
